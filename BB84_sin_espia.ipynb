```python
import numpy as np

rng = np.random.default_rng(seed=2026)

n =20

bits_alice = rng.integers(0, 2, size=n)
bases_alice = rng.integers(0,2, size=n)

print("Bits de alice:")
print(bits_alice)

print("Bases de Alice (0=Z, 1=X)")
print(bases_alice)

bases_bob = rng.integers(0,2, size = n)

print("Bases de Bob:")
print(bases_bob)

coinciden = bases_alice == bases_bob

bits_aleatorios = rng.integers(0,2, size =n)

bits_bob = np.where(coinciden, bits_alice, bits_aleatorios)


print("Bits medidos por Bob:")
print(bits_bob)

print("coinciden las bases?:")
print(coinciden)

clave_alice = bits_alice[coinciden]
clave_bob = bits_bob[coinciden]

print("clave cribada de Alice:")
print(clave_alice)

print("clave cribada de Bob:")
print(clave_bob)


errores = np.sum(clave_alice != clave_bob)
total_conservados = len(clave_alice)

if total_conservados > 0:
    qber = errores / total_conservados

    print("Bits conservados:", total_conservados)
    print("Errores:", errores)
    print(f"QBER: {qber * 100:.2f}%")
else:
    print("No se conservaron bits; no se puede calcular el QBER.")
```
